# Experimento: elegir k en k-means
**Hipótesis:** la inercia baja al subir k, pero el codo sugiere un compromiso.


```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
X, _ = make_blobs(n_samples=300, centers=3, random_state=0)
inertias = []
for k in range(1, 8):
    inertias.append(KMeans(n_clusters=k, n_init=10, random_state=0).fit(X).inertia_)
print('inertias', [round(v, 1) for v in inertias])
# codo grosero: mayor caída relativa temprana
drops = [inertias[i] - inertias[i+1] for i in range(len(inertias)-1)]
assert drops[0] > drops[-1]
```
